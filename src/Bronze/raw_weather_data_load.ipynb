```python
from pyspark.sql.functions import current_timestamp
```

```python
Table='project_mobility.bronze.weather_raw'

df=spark.read.format("csv").option("header", "true").option("inferSchema", "true").load("/Volumes/project_mobility/bronze/my_local_files/weather.csv")

df=df.withColumn("ingestion_timestamp",current_timestamp())

#display(df )

df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable(Table)
```
