```python
from pyspark.sql.functions import col, when, dayofweek, date_format, current_timestamp, lit
```

```python
src_Table='project_mobility.bronze.holidays_events_raw'
target_Table='project_mobility.silver.holidays_clean'
```

```python
df=spark.read.table(src_Table)

df_derived = df.withColumns({
    "is_holiday": lit(True)
    ,"impact_score": when(col("expected_impact") == "High",   3)
                     .when(col("expected_impact") == "Medium", 2)
                     .when(col("expected_impact") == "Low",    1)
                     .otherwise(0)
    ,"day_of_week":dayofweek(col("date"))
    ,"day_of_week_name":date_format(col("date"), "EEEE")
    ,"ingestion_timestamp": current_timestamp()
  })

df_derived.write.mode("overwrite").saveAsTable(target_Table)

#display(df_derived)
```
